## Lex only

```python
%{
    #include <stdio.h>
    #include <stlib.h>
    #incude <y.tab.h>
}%

%%

[0-9]+  {yylval = atoi(yytext); return NUMBER;}
[+]  {return ADD;}
[-]  {return SUB;}
[*]  {return MUL;}
[/]  {return DIV;}
[ \t\n]  ;
.  {return yytext[0];}
%%

int yywrap(){
    return 1;
}
```

```python
%{
    #include <stdio.h>
    int yylex();
    int yywrap();
%}


%token NUMBER
%token ADD SUB MUL DIV

%left ADD SUB
%left MUL DIV

%%

S : E  {printf("Result = %d\n",$1);}
E : E ADD E {$$=$1+$3;}|E SUB E{$$=$1-$3;}|E MUL E{$$=$1*$3;}|E DIV E{$$=$1/$3;}|'(' E ')'{$$=$2;}|NUMBER{$$=$1;};

%%

int main(){
    printf("Enter the expression : ");
    yyparse();
    return 0;
}

int yyerror(){}
```
